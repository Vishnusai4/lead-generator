## Optional: Query Database

````python
import sqlite3
import pandas as pd

# Connect to database
conn = sqlite3.connect('zendesk_leads_full.db')

# Example queries
print("Top 10 companies by employee count:\n")
query = """
SELECT domain, company_name, employees, zendesk_score
FROM leads
WHERE detected_zendesk = 1
ORDER BY employees DESC
LIMIT 10
"""
print(pd.read_sql_query(query, conn).to_string())

print("\n\nDetections by industry:\n")
query = """
SELECT industry, COUNT(*) as count
FROM leads
WHERE detected_zendesk = 1 AND industry IS NOT NULL
GROUP BY industry
ORDER BY count DESC
LIMIT 10
"""
print(pd.read_sql_query(query, conn).to_string())

conn.close()
````
